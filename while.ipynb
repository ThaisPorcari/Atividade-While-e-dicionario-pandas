```python
#Como funciona o While?
# while condição:

#exemplo 1 - Contar até onde o usuário pedir.
limite = int (input("informe um número limite: "))
inicio = 0
while inicio <= limite:
    print(inicio)
    inicio = inicio + 1
```

```python
#Exemplo 2
#Controle com while

senha_correta = "abc123"
tentativas = 0

while tentativas <3:
  senha = input("Digite sua senha: ")
  if senha_correta == senha:
    print("Bem vindo o sistema")
    break # o break quebra a estrutura de laço e finaliza esse laço
  else:
    print("Senha incorreta")
    tentativas = tentativas + 1
    #em python seria tentativas+=1, a forma de cima funciona em mais programas.
    print(f"Numero de tentativas = {tentativas}")
    if tentativas ==3:
      print("Senha bloqueada")
```

```python
#Exemplo 3
# While true -Enquanto o sistema for verdade (ou não der erro) ele continua .

while True:
  comando = input("Para sair, digite SAIR")
  if comando.lower() == "sair":
    print("Até mais")
    break
  else:
    print("Contnuando o sistema")
```

```python
#atividade 1 -Facil
#Quero um sistema que pergunte a um usuário se ele quer salvar uma nformação em uma lista de carros
#lista -->  carros = ["uo", "corola", "fusca"]
#Se o usuário quer colocar um novo carro, vai adicionar, se ele digitar sair, finaliza o sistema

carros = ["uno", "corola", "fusca"]
print("Bem vindo ao sistema de cadastro de carros")

while True:
    novo_carro = input("Digite um novo carro ou digite sair para finalizar o sistema?")
    if novo_carro.lower() == "sair":
      print("Sistema finalizado")
      break
    else:
      carros.append(novo_carro)
      for i in carros:
        print(i)
```
